```python
import requests

url = "https://power.larc.nasa.gov/api/temporal/daily/point"

params = {
    "parameters": "T2M",
    "community": "AG",
    "longitude": 11.36,
    "latitude": 8.89,
    "start": "20260101",
    "end": "20260501",
    "format": "JSON",
    "header": "false"
}

response = requests.get(url, params=params)

print("Status Code:", response.status_code)

```

```python
import pandas as pd

# 1. Translate NASA's JSON language into Python
data = response.json()

# 2. Go deep into the data to find the Temperature (T2M)
temp_data = data['properties']['parameter']['T2M']

# 3. Turn it into a neat table (DataFrame)
df = pd.DataFrame(list(temp_data.items()), columns=['Date', 'Temperature_C'])

# 4. Show the first 5 rows
df.head()
```

```python
import matplotlib.pyplot as plt

# Draw the line chart
plt.plot(df['Date'], df['Temperature_C'], color='red')

# Add titles so people know what they are looking at
plt.title('Jalingo Temperature (Jan - May 2026)')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')

# Rotate the dates so they don't overlap
plt.xticks(rotation=45)

# Show the chart
plt.show()
```
